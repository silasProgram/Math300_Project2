```python
def rand_graphs(v, j):
    e_total = v + j
    E = []
    for i in range(e_total):
        fact = False
        while(fact == False):
            w = random.randrange(0, 11); n = random.randrange(0, v); f = random.randrange(0, v)
            e_i = [w, n, f]
            if e_i[1] != e_i[2]:
                print(i)
                L = True
                if i == 0:
                    fact = True; continue
                for x in E:
                    if x[1] == e_i[1] and x[2] == e_i[2]:
                        L = False; continue 
                    if x[1] == e_i[2] and x[2] == e_i[1]:
                        L = False; continue 
                if L == True: fact = True
        E.append(e_i)   
    return E

print(rand_graphs(5,1))
```
