```python
def rand_graphs(v, j):
    e_total = v + j
    E = []
    for i in range(e_total):
        fact = False
        while(fact == False):
            w = random.randrange(1, 11); n = random.randrange(0, v); f = random.randrange(0, v)
            e_i = [w, n, f]
            if e_i[1] != e_i[2]: # This checks that the edge won't be a loop to a single vertex and restarts the loop if it is.
                L = True
                if i == 0: # This ensures that the first run passes the following tests automatically since there are no values in E.
                    fact = True; continue
                for x in E:   
                    if x[1] == e_i[1] and x[2] == e_i[2]: # This checks that there are no parallel edges. 
                        L = False; continue 
                    if x[1] == e_i[2] and x[2] == e_i[1]: # This checks that there are no isolated pairs.
                        L = False; continue 
                if L == True: fact = True
        E.append(e_i)   
    return E
v = 4; j = 1
E = rand_graphs(v, j)
print(E)
```

```python
adj = [[] for i in range(v)] # Initializes a 3D nested list, which we use to represent our graph for the algorithms.  
for x in E: # This loop groups all the vertices that are connected with a specific vertex into a 2D sublist in adj by appending the 1D sublist consisting of the vertex and weight.      
    adj[x[2]].append([x[0], x[1]])
    adj[x[1]].append([x[0], x[2]])
print(adj)
```

Actual algorithm.

```python
def dijkstra_algo(adj, source):
    vertices = len(adj) #number of vertices is the length of our adjacent list. 
    heap = [] #intiiallizing the heap as a list
    distance_ar = [float('inf')]*vertices #distance array, all distances are infinity until checked
    distance_ar[source]=0 #source distance to itself is zero
    heapq.heappush(heap, (0,source)) #placing our initial source in the heap 

    while heap: #this while loop iterates through the loop until nothing is within the heap. this works as throughout the while loop, things will stay within the heap until all edges have been mapped. 

        distance, vertex = heapq.heappop(heap) #distance is the total edge weight that the given vertex is away from the source. The popping pulls it out of the heap, as we're checking this value and evaluating the next vertices. 

        if distance> distance_ar[vertex]: #checks if we already have mapped this edge, if we have, it reruns the while loop. 
            continue 

        for x in adj[vertex]: #this iterates through all vertices and the edge weight connecting them to a given vertex via our adjacent matrix. 
            edge_weight = x[0];  vert_num= x[1]  

            if distance_ar[vert_num] > distance + edge_weight: #checks if the distance is not written yet. 
                distance_ar[vert_num] = distance + edge_weight #adds the total distance, so you have the distance from source to b 
                heapq.heappush(heap, (distance_ar[vert_num], vert_num)) #places the new vertex and distance in the heap to be checked. 
                print(distance_ar[vert_num], vert_num)

    return distance_ar

print(dijkstra_algo(adj, 0))
```
