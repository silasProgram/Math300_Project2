```python
def rand_graphs(v, j):
    e_total = v + j # Gives the total number of edges in the generated graph.  
    E = [] # Initializes the list of edges in the graph. 
    for i in range(e_total): # This loop generates every edge, appends them to the list of edges and returns the list.
        fact = False  
        while(fact == False): # This loop randomly generates the weight, two connected vertices, and then defines an edge by those values.
            w = random.randrange(1, 11); n = random.randrange(0, v); f = random.randrange(0, v) 
            e_i = [w, n, f]
            if e_i[1] != e_i[2]: # This checks that the edge won't be a loop to a single vertex and restarts the loop if it is.
                L = True
                if i == 0: # This ensures that the first run passes the following tests automatically since there are no values in E.
                    fact = True; continue
                for x in E:   
                    if x[1] == e_i[1] and x[2] == e_i[2]: # This checks that there are no parallel edges. 
                        L = False; continue 
                    if x[1] == e_i[2] and x[2] == e_i[1]: # This checks that there are no isolated pairs.
                        L = False; continue 
                if L == True: fact = True
        E.append(e_i)   
    return E
v = 4; j = 1
E = rand_graphs(v, j)
print(E)
```

```python
adj = [[] for i in range(v)] # Initializes a 3D nested list, which we use to represent our graph for the algorithms.  
for x in E: # This loop groups all the vertices that are connected with a specific vertex into a 2D sublist in adj by appending the 1D sublist consisting of the vertex and weight.      
    adj[x[2]].append([x[0], x[1]])
    adj[x[1]].append([x[0], x[2]])
print(adj)
```
